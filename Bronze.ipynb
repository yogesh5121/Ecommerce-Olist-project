**Reading the data from source files**

```python
df_geo = spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/geo_location")
```

```python
df_order = spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/orders")
```

```python
df_reviews =spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/orders_reviews")
```

```python
df_payment = spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .csv("/Volumes/olist/bronze/payments")
```

```python
df_product =spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/product")


```

```python
df_product_cat = spark.read.option("header",True)\
                   .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/product_category")
```

```python
df_seller = spark.read.option("header",True)\
                 .option("inferschema",True)\
                     .parquet("/Volumes/olist/bronze/seller")
```

```python
df_customer = spark.read.option("header",True)\
                        .option("inferschema",True)\
                        .csv("/Volumes/olist/bronze/customer")

df_customer.display()
```

```python
df_order_item = spark.read.option("header",True)\
                          .option("inferschema",True)\
                          .csv("/Volumes/olist/bronze/order_items")

df_order_item.display()
```

Converting the raw files into delta tables

```python
spark.sql("DROP TABLE olist.bronze.olist_payment")
```

```python
df_order_item.write.mode("overwrite").saveAsTable("olist.bronze.olist_order_item")
```

```python
df_customer.write.mode("overwrite").saveAsTable("olist.bronze.olist_customer")
```

```python
df_payment.write.mode("overwrite").saveAsTable("olist.bronze.olist_payment")
```

```python
df_geo.write.mode("overwrite").saveAsTable("olist.bronze.olist_geo")
df_order.write.mode("overwrite").saveAsTable("olist.bronze.olist_order")
df_reviews.write.mode("overwrite").saveAsTable("olist.bronze.olist_reviews")

df_product.write.mode("overwrite").saveAsTable("olist.bronze.olist_product")
df_product_cat.write.mode("overwrite").saveAsTable("olist.bronze.olist_product_category")
df_seller.write.mode("overwrite").saveAsTable("olist.bronze.olist_seller")
```
