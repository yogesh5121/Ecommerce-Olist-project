```python
df_seller = spark.read.table("olist.bronze.olist_seller")

df_seller.display()
```

```python
from pyspark.sql import functions as F

# Clean column names by removing double quotes
for old_col in df_seller.columns:
    new_col = old_col.replace('"', '')
    df_seller = df_seller.withColumnRenamed(old_col, new_col)

# Remove double quotes from all string column values
for col in df_seller.columns:
    df_seller = df_seller.withColumn(col, F.regexp_replace(F.col(col), '"', ''))

df_seller.display()
```

```python
df_seller.write.format("delta").mode("overwrite").saveAsTable("olist.silver.seller")
```
