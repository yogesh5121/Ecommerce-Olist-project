```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
df_reviews = spark.read.table("olist.bronze.olist_reviews")
```

```python
timestamp_columns = [
    "review_creation_date",
    "review_answer_timestamp\r"
]
for column_name in timestamp_columns:
    df_reviews = df_reviews.withColumn(
        column_name, 
        F.when(F.col(column_name) == "", None).otherwise(F.to_timestamp(F.col(column_name)))
    )
```

```python
df_reviews.display()
```

```python
df_reviews.write.format("delta").mode("overwrite").saveAsTable("olist.silver.olist_reviews")
```
