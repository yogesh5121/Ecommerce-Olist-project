```python
df_geo = spark.read.table("olist.bronze.olist_geo")
df_geo.display()
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import DoubleType

# Convert lat/lng to double type and rename columns
df_geo = df_geo.withColumn("geolocation_lat", col("geolocation_lat").cast(DoubleType())) \
               .withColumn("geolocation_lng", col("geolocation_lng").cast(DoubleType())) \
               .withColumnRenamed("geolocation_lat", "geolocation_latitude") \
               .withColumnRenamed("geolocation_lng", "longitude")

df_geo.display()
```

```python
df_geo.write.mode("overwrite").format("delta").saveAsTable("olist.silver.olist_geo")
```
